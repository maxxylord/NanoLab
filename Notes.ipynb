##### **Task 01**
After initially setting up the measurement the unfiltered voltage level on the detecto reached $V_{\mathrm{init, RMS}} = 9.11 V$ <br>
The noise level of the room was around $\sigma_V = 61.0 \mathrm{mV}$<br>
After implementing the polarizers, we turned the first polarizer, since the laser output is somwhat linearly polarized, so we seeked the best alignment of the first filter with it.<br>

```python
angles = np.linspace(0, 180, 19) # in degrees
print(angles)
voltage= np.array([1.392, 1.185, 0.934, 0.690, 0.470, 0.221, 0.126, 0.058, .064, .135, .347, .618, .830, 1.107, 1.339, 1.435, 1.553, 1.498, 1.364]) # in V
plt.figure()
plt.scatter(angles, voltage)
plt.title('Magnus law between two linear polarization filters')
plt.xlabel(r'$\mathrm{Angle}(^{\circ})$')
plt.ylabel('RMS Voltage (V)')

min_angle = angles[np.argmin(voltage)]
max_angle = angles[np.argmax(voltage)]

plt.axvline(x=min_angle, color='red', linestyle='--', label='Perpendicular')
plt.axvline(x=max_angle, color='green', linestyle='--', label='Parallel')
plt.legend()

plt.show()
```

##### **Task 02**
Starting from the second task we have closed the shaders on the window. The new noise level is $\sigma_V = 50\mathrm{mV}$<br>
As the previous measurement suggests the parallel orientation of the two filters was at an angle of $160^{\circ}$<br>
With parallel polarizers the maximal intensity was detected at exactly (with readoff precision, of $3^{\circ}$) $0^{\circ}$ of LC rotation.<br>
After setting the polarizers perpendicular the new maximal LC orientation was indeed at $45^{\circ}$ <br>
Measuring the detector voltage we realised a constatnt voltage level irrespective of the analyzer orientation (within margin of error), which can only mean that the LC circuarly polarised the linear beam.

```python
LCangle = np.linspace(0, 180, 19)

LCvoltage = np.array([.451, .471, .501, .510, .505, .512, .499, .480, .461, .431, .425, .425, .415, .415, .423, .420, .458, .480, .496])
avg = np.average(LCvoltage)
std = np.std(LCvoltage)

theta = np.deg2rad(LCangle)

plt.figure(figsize=(7, 5))
ax = plt.subplot(111, projection='polar')
bx = plt.subplot(111, projection='polar')

ax.plot(theta, LCvoltage, marker='o', linestyle='-', color='b', label='LC implemented')
bx.plot(theta, voltage, marker='o', linestyle='-', color='r', label='LC missing')

ax.set_thetamin(0)
ax.set_thetamax(360)
plt.legend()

plt.title("Intensity of output light with and without LC implemented")

plt.show()

plt.figure(figsize=(7, 5))
plt.scatter(angles, voltage, color='r', label='LC implemented')
plt.scatter(LCangle, LCvoltage, color='b', label='LC missing')
plt.hlines([avg - std,avg, avg + std], np.min(LCangle), np.max(LCangle), color='g', linestyles='dashed', label='Average', alpha=.5)
plt.hlines([avg - std, avg + std],np.min(LCangle), np.max(LCangle), color = '#FF007F', linestyles='dashed', label='Standard deviation', alpha=1)
plt.legend()
plt.grid()
plt.show()
```

```python
angles = np.array([-10, -5, 0, 5, 10])*-1
bessel_two = np.array([0.508241, 0.467995, 0.421146, 0.376127, 0.335960])
slope, intercept = np.polyfit(angles, bessel_two, 1)
k = slope*angles + intercept
print(f"Slope: {slope}")
plt.figure()
plt.scatter(angles, bessel_two)
plt.grid()
plt.xlabel('Angle around 45 (deg)')
plt.ylabel('2f peak(uV)')
plt.plot(angles, k, c= 'r', label = f'Linear fit slope = {slope}')
plt.title("40Hz 500mV modulation")
plt.legend()
plt.show()
```
